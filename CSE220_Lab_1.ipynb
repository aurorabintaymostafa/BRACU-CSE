```python
#Task 01: Merge Lineup

def mergeLineup(pokemon_1, pokemon_2):
  #To Do
  result_array = np.zeros(len(pokemon_1),dtype = int)
  for i in range(len(pokemon_1)):
    x = pokemon_1[i]
    if x == None:
      x = 0
    y = pokemon_2[len(pokemon_1)-1-i]
    if y == None:
      y = 0
    result_array[i] = x + y
  return result_array


print("///  Task 01: Merge Lineup  ///")
pokemon_1 = np.array([12, 3, 25, 1, None])
pokemon_2 = np.array([5, -9, 3, None, None] )
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [12, 3, 28, -8, 5]
unittest.output_test(returned_value, np.array([12, 3, 28, -8, 5]))
pokemon_1 = np.array([4, 5, -1, None, None])
pokemon_2 = np.array([2, 27, 7, 12, None])
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [4,17,6,27,2]
unittest.output_test(returned_value, np.array([4,17,6,27,2]))
```

```python
from typing_extensions import TypeVarTuple
# Task 02: Discard Cards

def discardCards(cards, t):
  # TO DO
  found = True
  i = 0
  while i < len(cards):
    if cards[i] == t and found == True:
      for j in range(i+1, len(cards)):
        cards[j-1] = cards[j]
      cards[len(cards)-1] = 0
      found = False
    else:
      if cards[i] == t:
       found = True
      i += 1
  return cards




print("///  Task 02: Discard Cards  ///")
cards = np.array([1,3,7,2,5,2,2,2,0])
returned_value = discardCards(cards, 2)
print(f'Task 2: {returned_value}') # This should print [1,3,7,5,2,2,0,0,0]
unittest.output_test(returned_value, np.array([1,3,7,5,2,2,0,0,0]))

cards = np.array([5,5,5,0,0])
returned_value = discardCards(cards, 5)
print(f'Task 2: {returned_value}') # This should print [5,0,0,0,0]
unittest.output_test(returned_value, np.array([5,0,0,0,0]))
```

```python
#Task 03: Decryption Process

def decrypt_matrix(matrix):
  #To Do
  row, col = matrix.shape
  new_array = np.zeros(col, dtype = int)
  result = np.zeros(col-1,dtype = int)
  for i in range(col):
    sum = 0
    for j in range(row):
      sum += matrix[j][i]
    new_array[i] = sum
  for i in range(col-1):
    result[i] = new_array[i] - new_array[i+1]
  return result


matrix=np.array([[1,3,1],
                 [6,4,2],
                 [5,1,7],
                 [9,3,3],
                 [8,5,4]
                 ])

returned_array=decrypt_matrix(matrix)
print(returned_array)
#This should print [-13, 1]
```

```python
#Task 06: Matrix Compression

def compress_matrix(matrix):
  #TO DO
  row, col = matrix.shape
  row = row//2
  col = col//2
  new_array = np.zeros((row,col), dtype = int)
  #returned_array = np.zeros(col-1, dtype = int)
  for i in range(row):
    for j in range(col):
      temp = matrix[2*i][2*j] + matrix[2*i][2*j+1] + matrix[2*i+1][2*j] + matrix[2*i+1][2*j+1]
      new_array[i][j] = temp
  return new_array




matrix=np.array([[1,2,3,4],
                 [5,6,7,8],
                 [1,3,5,2],
                 [-2,0,6,-3]
                 ])
print_matrix(matrix)
print("...............")
returned_array=compress_matrix(matrix)
print_matrix(returned_array)
#This should print

#|  14  |  22 |
#--------------
#|  2  |  10  |
#--------------
```

```python
#Task 07: Game Arena

def play_game(arena):
  #TO DO
  row, col = arena.shape
  player_point = 0
  for i in range(row):
    for j in range(col):
      if arena[i][j] % 50 == 0 and arena[i][j] != 0:
        if 0 < i+1 < row and arena[i+1][j] == 2:
          player_point += 2
        if row > i-1 >= 0 and arena[i-1][j] == 2:
          player_point += 2
        if 0 <= j+1 < col and  arena[i][j+1] == 2:
          player_point += 2
        if col > j-1 >= 0 and arena[i][j-1] == 2:
          player_point += 2
  if player_point < 10 :
    print(f'Points Gained: {player_point} Your team is out. ')
  else:
    print(f'Points Gained: {player_point} Your team has survived the game. ')

arena=np.array([[0,2,2,0],
                [50,1,2,0],
                [2,2,2,0],
                [1,100,2,0]
                ])
print_matrix(arena)
play_game(arena)
#This should print
#Points Gained: 6. Your team is out.

print(".....................")
arena=np.array([[0,2,2,0,2],
                [1,50,2,1,100],
                [2,2,2,0,2],
                [0,200,2,0,0]
                ])
print_matrix(arena)
play_game(arena)
#This should print
#Points Gained: 14. Your team has survived the game.
```
